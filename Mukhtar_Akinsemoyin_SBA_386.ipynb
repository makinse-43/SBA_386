```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read Data from the woocommerce-product-export.csv
df = pd.read_csv('data/woocommerce-product-export.csv')
```

```python
# Show a concise summary of the columns using the info() method.
df.info()
```

```python
# Show a summary of statistics pertaining to the columns.
df.describe()
```

```python
# Print the first five rows by default.
df.head()
```

```python
# Print the last five rows by default.
df.tail()
```

```python
# Print  the "total_profit" and “month_number” columns only.
df[['total_profit', 'month_number']]
```

```python
# Read the total profit of all months and show it using the Bar plot. 
    # The generated Bar plot must include the following properties: 
        # X label name = Month number.
        # Y label name = Total profit.
        # Title = Company profit per month.

month_number = df['month_number']
total_profit = df['total_profit']

plt.bar(month_number, total_profit)

plt.title('Company Profit per Month')
plt.xlabel('Month Number')
plt.ylabel('Total Profit')

plt.yticks(range(100000, 500001, 100000))
plt.grid(alpha=0.5, linestyle='--')

plt.show()
```

```python
# Read the total profit of all of the months, and show the line plot with the following style properties:
    # The generated line plot must include the following Style properties: 
        # The line style should be dotted and the line color should be red.
        # Show the legend at the lower-right location.
        # X label name = Month number
        # Y label name = Total profit.
        # Title = Company profit per month.
        # Add a circle marker; the line marker color = black.
        # The line width should be 3.

plt.plot(month_number, total_profit, linestyle='dotted', color='red', linewidth=3, marker='o', markerfacecolor='black', markersize=8)

plt.title('Company Profit per Month')
plt.xlabel('Month Number') 
plt.ylabel('Total Profit')
plt.legend(['Profit Data of Last Year'], loc='lower right')
plt.xticks(range(1, 13))
plt.yticks(range(100000, 500001, 100000))

plt.show()
```

```python
# Print all of the product sales data and show it using a multi-line plot.
    # Display the number of units sold per month for each product using multi-line plots. (i.e., show a separate plotline for each product).

print(df)
```

```python
# Define data

facecream = df['facecream']
facewash = df['facewash']
toothpaste = df['toothpaste']
bathingsoap = df['bathingsoap']
shampoo = df['shampoo']
moisturizer = df['moisturizer']
```

```python
# Display the number of units sold per month for each product using multi-line plots. (i.e., show a separate plotline for each product).

plt.figure(figsize=(10, 8))

plt.plot(month_number, facecream, label='Face Cream Sales Data', marker='o')
plt.plot(month_number, facewash, label='Face Wash Sales Data', marker='s')
plt.plot(month_number, toothpaste, label='Toothpaste Sales Data', marker='^')
plt.plot(month_number, bathingsoap, label='Bathing Soap Sales Data', marker='D')
plt.plot(month_number, shampoo, label='Shampoo Sales Data', marker='v')
plt.plot(month_number, moisturizer, label='Moisturizer Sales Data', marker='p')

plt.xlabel('Month Number')
plt.ylabel('Sales Units in Number')
plt.title('Sales Data')
plt.legend(loc='upper left')
plt.xticks(range(1, 13))
plt.yticks([1000, 2000, 4000, 6000, 8000, 10000, 12000, 15000, 18000])

plt.show()
```

```python
# Read “bathingsoap” sales data for each month and show it using a scatter plot.
    # Add a grid to the plot. The grid line style should be “–“.

plt.scatter(month_number, bathingsoap, label='Bathing Soap Sales Data', marker='o')
plt.xlabel('Month Number')
plt.ylabel('Number of Units Sold')
plt.title('Bathing Soap Sales Data')

plt.legend(loc='upper left')
plt.xticks(range(1, 13))
plt.yticks(range(6000, 14001, 2000))
plt.grid(linestyle='--')

plt.show()
```

```python
# Given Dataset:
height=[121.9,124.5,129.5,134.6,139.7,147.3,152.4,157.5,162.6]
weight=[19.7,21.3,23.5,25.9,28.5,32.1,35.7,39.6,43.2]
```

```python
# Show the average weight against the average height (use a line chart or any other chart):
    # The generated line plot must include the following Style properties:
        # The line style should be dash-dot and the line color should be green.
        # Show the legend at the lower-right location.
        # X label name = Weight in kg.
        # Y label name = Height in cm.
        # Title = Average weight with respect to average height.
        # Add a circle marker: marker size = 10; circle marker color = green.

plt.plot(weight, height, linestyle='-.', color='green', marker='o', markersize=10, markerfacecolor='green')

plt.xlabel('Weight in kg')
plt.ylabel('Height in cm')
plt.title('Average Weight With Respect to Average Height')
plt.legend(['Average Weight by Average Height'], loc='lower right')

plt.show()
```
